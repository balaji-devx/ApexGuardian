# Bengaluru Traffic Modeling & Feature Engineering
This notebook defines data cleaning, feature transformations, and segment speed modeling for Bengaluru road corridors.

```python
import numpy as np
import pandas as pd

def generate_raw_traffic_samples(n=5000):
    np.random.seed(42)
    road_type = np.random.choice([1, 2, 3], size=n, p=[0.25, 0.50, 0.25])
    time_of_day = np.random.randint(0, 24, size=n)
    day_of_week = np.random.randint(0, 7, size=n)
    weather_condition = np.random.choice([0, 1], size=n, p=[0.8, 0.2])
    distance_km = np.random.uniform(0.5, 8.0, size=n)
    historical_avg_speed = np.where(road_type == 1, 55.0, np.where(road_type == 2, 35.0, 22.0)) + np.random.normal(0, 3, size=n)
    
    actual_speed = []
    for i in range(n):
        factor = 1.0
        if (8 <= time_of_day[i] <= 10 or 17 <= time_of_day[i] <= 20) and day_of_week[i] < 5:
            factor *= np.random.uniform(0.35, 0.55)
        elif day_of_week[i] >= 5:
            factor *= np.random.uniform(0.85, 1.05)
        else:
            factor *= np.random.uniform(0.75, 0.95)
        if weather_condition[i] == 1:
            factor *= 0.75
        actual_speed.append(max(6.0, historical_avg_speed[i] * factor + np.random.normal(0, 2.5)))
        
    df = pd.DataFrame({
        'road_type': road_type,
        'time_of_day': time_of_day,
        'day_of_week': day_of_week,
        'historical_avg_speed': historical_avg_speed,
        'weather_condition': weather_condition,
        'distance_km': distance_km,
        'actual_speed': actual_speed
    })
    return df

df = generate_raw_traffic_samples()
df.to_csv('cleaned_traffic.csv', index=False)
print('Dataset generated with shape:', df.shape)
```
